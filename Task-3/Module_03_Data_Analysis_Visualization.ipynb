## 1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
print("Libraries imported successfully.")
```

## 2. Create and load a sample sales dataset

The dataset deliberately contains a missing value and a duplicate row to demonstrate data cleaning.

```python
data = {
    "Date": ["2026-01-05","2026-01-12","2026-01-19","2026-02-03","2026-02-10",
             "2026-02-17","2026-03-02","2026-03-09","2026-03-16","2026-03-23",
             "2026-04-06","2026-04-13","2026-04-20","2026-04-27","2026-03-09"],
    "Product": ["Laptop","Phone","Headphones","Laptop","Phone","Tablet","Headphones",
                "Laptop","Phone","Tablet","Laptop","Headphones","Phone","Tablet","Laptop"],
    "Units_Sold": [2,5,8,3,4,np.nan,10,1,6,3,2,7,5,4,1],
    "Unit_Price": [55000,18000,2500,55000,18000,22000,2500,55000,18000,22000,
                   55000,2500,18000,22000,55000],
    "Region": ["South","North","East","West","South","North","East","West","South",
               "North","East","West","South","North","West"],
    "Customer_Email": ["a@example.com","b@example.com","c@example.com","d@example.com",
                       "e@example.com","f@example.com","g@example.com","h@example.com",
                       "i@example.com","j@example.com","k@example.com","l@example.com",
                       "m@example.com",None,"h@example.com"]
}
df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])
print(f"Rows: {df.shape[0]} | Columns: {df.shape[1]}")
display(df.head())
```

## 3. Explore the dataset

```python
print("Dataset information:")
df.info()
print("\nSummary statistics:")
display(df.describe(include="all"))
print("\nMissing values per column:")
display(df.isna().sum())
```

## 4. Clean the data

Remove duplicate records, fill missing unit counts with the median, and label missing emails as `Unknown`.

```python
rows_before = len(df)
clean_df = df.drop_duplicates().copy()
duplicates_removed = rows_before - len(clean_df)
clean_df["Units_Sold"] = clean_df["Units_Sold"].fillna(clean_df["Units_Sold"].median()).astype(int)
clean_df["Customer_Email"] = clean_df["Customer_Email"].fillna("Unknown")

print(f"Duplicate rows removed: {duplicates_removed}")
print("Missing values after cleaning:")
display(clean_df.isna().sum())
display(clean_df.head())
```

## 5. Analyze sales with Pandas and NumPy

```python
clean_df["Revenue"] = clean_df["Units_Sold"] * clean_df["Unit_Price"]
clean_df["Month"] = clean_df["Date"].dt.to_period("M").astype(str)

total_revenue = clean_df["Revenue"].sum()
print(f"Total revenue: ₹{total_revenue:,.2f}")
print(f"Total units sold: {clean_df['Units_Sold'].sum()}")

product_sales = clean_df.groupby("Product", as_index=False)["Revenue"].sum().sort_values("Revenue", ascending=False)
region_sales = clean_df.groupby("Region", as_index=False)["Revenue"].sum()
monthly_sales = clean_df.groupby("Month", as_index=False)["Revenue"].sum().sort_values("Month")

print("Revenue by product:")
display(product_sales)
print("Revenue by region:")
display(region_sales)
print("Revenue by month:")
display(monthly_sales)

revenue_array = clean_df["Revenue"].to_numpy()
print(f"NumPy mean revenue: ₹{np.mean(revenue_array):,.2f}")
print(f"NumPy median revenue: ₹{np.median(revenue_array):,.2f}")
```

## 6. Bar chart: Revenue by product

```python
plt.figure(figsize=(8,5))
plt.bar(product_sales["Product"], product_sales["Revenue"])
plt.title("Total Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.ticklabel_format(style="plain", axis="y")
plt.tight_layout()
plt.show()
```

## 7. Line chart: Monthly revenue trend

```python
plt.figure(figsize=(8,5))
plt.plot(monthly_sales["Month"], monthly_sales["Revenue"], marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.ticklabel_format(style="plain", axis="y")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 8. Pie chart: Revenue share by product

```python
plt.figure(figsize=(7,7))
plt.pie(product_sales["Revenue"], labels=product_sales["Product"], autopct="%1.1f%%", startangle=90)
plt.title("Revenue Share by Product")
plt.tight_layout()
plt.show()
```

## 9. Key findings and conclusion

```python
top_product = product_sales.iloc[0]
top_region = region_sales.sort_values("Revenue", ascending=False).iloc[0]
best_month = monthly_sales.sort_values("Revenue", ascending=False).iloc[0]

print(f"Highest-revenue product: {top_product['Product']} (₹{top_product['Revenue']:,.2f})")
print(f"Highest-revenue region: {top_region['Region']} (₹{top_region['Revenue']:,.2f})")
print(f"Highest-revenue month: {best_month['Month']} (₹{best_month['Revenue']:,.2f})")
print(f"Total revenue: ₹{total_revenue:,.2f}")
print(f"Duplicate rows removed: {duplicates_removed}")
```

### Conclusion
Pandas and NumPy were used to explore, clean, and analyze sales data. Matplotlib charts compared revenue by product, showed monthly trends, and displayed product revenue shares.
